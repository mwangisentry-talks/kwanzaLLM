```python

import tensorflow as tf
import numpy as np  

# Define matrices
matrix1 = np.array([(2, 2, 2), (2, 2, 2), (2, 2, 2)], dtype='int32')
matrix2 = np.array([(1, 1, 1), (1, 1, 1), (1, 1, 1)], dtype='int32')
print("Matrix 1:")
print(matrix1)
print("Matrix 2:")
print(matrix2)

# Convert numpy arrays to TensorFlow tensors
matrix1 = tf.constant(matrix1)
matrix2 = tf.constant(matrix2)

# Compute matrix product and sum
matrix_product = tf.matmul(matrix1, matrix2)
matrix_sum = tf.add(matrix1, matrix2)

# Define another matrix for determinant calculation
matrix_3 = np.array([(2, 7, 2), (1, 4, 2), (9, 0, 2)], dtype='float32')
print("Matrix 3:")
print(matrix_3)

# Compute determinant
matrix_det = tf.linalg.det(matrix_3)

# Print results
print("Matrix Product:")
print(matrix_product.numpy())
print("Matrix Sum:")
print(matrix_sum.numpy())
print("Determinant of Matrix 3:")
print(matrix_det.numpy())
```
