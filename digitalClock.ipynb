```python
import tkinter as tk

from time import strftime

rt = tk.Tk()
rt.title("Digital Clock")

clockLabel = tk.Label(rt, font=("Helvetica", 48), bg="Green", fg="Yellow")
clockLabel.pack(anchor="center", fill="both", expand="True")

def update_time():
    currenttime = strftime("%H:%M:%S")
    clockLabel.config(text=currenttime)
    clockLabel.after(1000,update_time)
update_time()
rt.mainloop()

```

```python
import whois

def get_domain(domain_name):
    try:
        # Fetch WHOIS data
        domain = whois.query(domain_name)

        # Print WHOIS information
        if domain:
            print(f"Domain Name: {domain.name}")
            print(f"Registrar: {domain.registrar}")
            print(f"Creation Date: {domain.creation_date}")
            print(f"Expiration Date: {domain.expiration_date}")
            print(f"Last Updated: {domain.last_updated}")
            print(f"Name Servers: {', '.join(domain.name_servers) if domain.name_servers else 'N/A'}")
        else:
            print(f"WHOIS information for '{domain_name}' not found.")
    except Exception as e:
        print(f"Error: Unable to fetch WHOIS information. {e}")

if __name__ == "__main__":
    domain = input("Enter Domain Name: ")
    get_domain(domain)
```
