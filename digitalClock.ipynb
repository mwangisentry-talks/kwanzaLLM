```python
import tkinter as tk

from time import strftime

rt = tk.Tk()
rt.title("Digital Clock")

clockLabel = tk.Label(rt, font=("Helvetica", 48), bg="Green", fg="Yellow")
clockLabel.pack(anchor="center", fill="both", expand="True")

def update_time():
    currenttime = strftime("%H:%M:%S")
    clockLabel.config(text=currenttime)
    clockLabel.after(1000,update_time)
update_time()
rt.mainloop()

```
